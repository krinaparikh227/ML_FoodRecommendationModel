```python
import pandas as pd
```

```python
recipes=pd.read_csv("recipes.csv")
```

```python
#clean the columns
recipes.columns = recipes.columns.str.strip().str.lower()
```

```python
food_metadata = recipes[[
    'recipeid',
    'name',
    'recipeingredientparts',
    'calories',
    'proteincontent',
    'fatcontent',
    'carbohydratecontent',
    'recipecategory',
    'totaltime',
    'recipeinstructions'
]].copy()
```

```python
# Rename columns (IMPORTANT)
food_metadata.columns = [
    'item_id',
    'recipe_name',
    'ingredients',
    'calories',
    'protein',
    'fat',
    'carbs',
    'category',
    'prep_time',
    'instructions'
]
```

```python
# Handle missing values
food_metadata.fillna({
    'ingredients': '',
    'category': 'Unknown',
    'prep_time': 0,
    'instructions': ''
}, inplace=True)
```

```python
# Convert numeric columns
for col in ['calories', 'protein', 'fat', 'carbs', 'prep_time']:
    food_metadata[col] = pd.to_numeric(food_metadata[col], errors='coerce').fillna(0)
```

```python
# Save file
food_metadata.to_csv("food_metadata_clean.csv", index=False)
```
