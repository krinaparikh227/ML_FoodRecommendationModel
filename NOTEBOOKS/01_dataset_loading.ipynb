```python
# Clean column names
recipes.columns = recipes.columns.str.lower().str.strip()
reviews.columns = reviews.columns.str.lower().str.strip()

print("Recipes columns:", recipes.columns.tolist())
print("Reviews columns:", reviews.columns.tolist())

# Auto-detect recipe column
possible_recipe_cols = ['id', 'recipe_id', 'recipeid', 'rid','RecipeId']

recipe_col = None
for col in possible_recipe_cols:
    if col in recipes.columns:
        recipe_col = col
        break

if recipe_col is None:
    raise Exception("No recipe ID column found. Check column names manually.")

# Auto-detect user column
possible_user_cols = ['user_id', 'userid', 'uid','ReviewId','authorid']

user_col = None
for col in possible_user_cols:
    if col in reviews.columns:
        user_col = col
        break

if user_col is None:
    raise Exception("No user ID column found.")

print("Detected recipe column:", recipe_col)
print("Detected user column:", user_col)

# Final stats
print("Total Recipes:", recipes[recipe_col].nunique())
print("Total Users:", reviews[user_col].nunique())
print("Total Reviews:", len(reviews))
```
